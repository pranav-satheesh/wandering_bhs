```python
with h5py.File("/orange/lblecha/pranavsatheesh/wandering_bhs/Data/wandering_bh_SM5_DFD_3_TNG.hdf5") as f:
    is_wandering = f["bh/is_wandering"][:]
    r_kpc = f["bh/r_kpc"] # all radii
    wandering_r_kpc = f["bh/r_kpc"][is_wandering]          # distances of just the wandering BHs
    nonwander_r_kpc = f["bh/r_kpc"][np.invert(is_wandering)]      # distances of all other BHs
    lum = f["bh/lbol_erg_s"] # all lum
    wandering_lum = f["bh/lbol_erg_s"][is_wandering]       # luminosities of wanderers
    nonwander_lum = f["bh/lbol_erg_s"][np.invert(is_wandering)]    # luminosities of others

    snapchoice = 7                                                     # CHANGE SNAPSHOT HERE
    num_halos_in_snap = f["snapshot/n_subhalos_with_bh"][snapchoice]
    subhalochoice = 2 % num_halos_in_snap                               # CHANGE SUBHALO HERE
    subhalo_start = f["snapshot/subhalo_start"][snapchoice]
    bh_start = f["subhalo/bh_start"][subhalo_start+subhalochoice]
    halfmass_rad = f["subhalo/stellar_halfrad_kpc"][subhalo_start+subhalochoice]
    num_bh_in_halo = f["subhalo/bh_count"][subhalo_start+subhalochoice]
    wandering_in_halo = is_wandering[bh_start:bh_start+num_bh_in_halo]
    r_in_halo = r_kpc[bh_start:bh_start+num_bh_in_halo]
    r_wand_in_halo = r_in_halo[wandering_in_halo]
    r_nonwand_in_halo = r_in_halo[np.invert(wandering_in_halo)]
    lum_in_halo = lum[bh_start:bh_start+num_bh_in_halo]
    lum_wand_in_halo = lum_in_halo[wandering_in_halo]
    lum_nonwand_in_halo = lum_in_halo[np.invert(wandering_in_halo)]
```

```python
# To plot BH mass (as fraction of halo mass) against radius in halo
with h5py.File("/orange/lblecha/pranavsatheesh/wandering_bhs/Data/wandering_bh_SM5_DFD_3_TNG.hdf5") as f:
    is_wandering = f["bh/is_wandering"][:]
    r_kpc = f["bh/r_kpc"] # all radii
    mass = f["bh/mass_msun"] # all masses
    
    snapchoice = 19                                           # CHANGE SNAPSHOT HERE
    num_halos_in_snap = f["snapshot/n_subhalos_with_bh"][snapchoice]
    print("Number of halos in snapshot:", num_halos_in_snap)
    for subhalochoice in range(num_halos_in_snap):
        if subhalochoice % 100 == 0: print(f"processing subhalo {subhalochoice}")
        subhalo_start = f["snapshot/subhalo_start"][snapchoice]
        bh_start = f["subhalo/bh_start"][subhalo_start+subhalochoice]
        halo_mass = f["subhalo/subhalo_total_mass_msun"][subhalo_start+subhalochoice]
        halfmass_rad = f["subhalo/stellar_halfrad_kpc"][subhalo_start+subhalochoice]
        num_bh_in_halo = f["subhalo/bh_count"][subhalo_start+subhalochoice]
        wandering_in_halo = is_wandering[bh_start:bh_start+num_bh_in_halo]
        r_in_halo = r_kpc[bh_start:bh_start+num_bh_in_halo]
        bh_mass_in_halo = mass[bh_start:bh_start+num_bh_in_halo]
        mass_frac = bh_mass_in_halo/halo_mass
        plt.scatter(np.log10(r_in_halo), np.log10(mass_frac)) #, label=str(subhalochoice))
        plt.axvline(np.log10(halfmass_rad), color="k")
#plt.legend()
plt.xlabel("log10(radius [kpc])")
plt.ylabel("log10(bh mass as fraction of halo)")
plt.title(f"BH Mass vs Rad for subhalo of snap {snapchoice}") 
```

```python
# To plot BH mass (as fraction of halo mass) against radius in halo. DISCOUNTS EXTRACENTRALS
def bh_mass_frac_vs_rad_noncentral(snapchoice, subhalo=-1, markersize=20):
    with h5py.File("/orange/lblecha/pranavsatheesh/wandering_bhs/Data/wandering_bh_SM5_DFD_3_TNG.hdf5") as f:
        # is_wandering = f["bh/is_wandering"][:]
        r_kpc = np.array(f["bh/r_kpc"]) # all radii
        isnt_extracentral = r_kpc > 1e-8
        mass = f["bh/mass_msun"] # all masses

        num_halos_in_snap = f["snapshot/n_subhalos_with_bh"][snapchoice]
        if subhalo==-1:
            print("Number of halos in snapshot:", num_halos_in_snap)
            print("Processing subhalo:")
            subarr = range(num_halos_in_snap)
        else: subarr = [subhalo%num_halos_in_snap]
        for subhalochoice in subarr:
            if subhalochoice % 200 == 0: print(subhalochoice, end=" ")
            subhalo_start = f["snapshot/subhalo_start"][snapchoice]
            bh_start = f["subhalo/bh_start"][subhalo_start+subhalochoice]
            halo_mass = f["subhalo/subhalo_total_mass_msun"][subhalo_start+subhalochoice]
            halfmass_rad = f["subhalo/stellar_halfrad_kpc"][subhalo_start+subhalochoice]
            num_bh_in_halo = f["subhalo/bh_count"][subhalo_start+subhalochoice]
            valid_in_halo = isnt_extracentral[bh_start:bh_start+num_bh_in_halo]
            # wandering_in_halo = is_wandering[bh_start:bh_start+num_bh_in_halo]
            r_in_halo = (r_kpc[bh_start:bh_start+num_bh_in_halo])[valid_in_halo]
            if r_in_halo.size == 0: continue
            bh_mass_in_halo = (mass[bh_start:bh_start+num_bh_in_halo])[valid_in_halo]
            mass_frac = bh_mass_in_halo/halo_mass
            plt.scatter(np.log10(r_in_halo), np.log10(mass_frac), s=markersize) #, label=str(subhalochoice))
            # plt.axvline(np.log10(halfmass_rad), color="k") # better for individual halos
    #plt.legend()
    plt.xlabel("log10(radius [kpc])")
    plt.ylabel("log10(bh mass as fraction of halo)")
    subtitle = "subhalo " + str(subhalo) if subhalo != -1 else "all subhalos"
    title = "BH Mass vs Rad for " + subtitle + " of snap " + str(snapchoice) 
    plt.title(title) 
    plt.plot()
```

```python
# BH mass (as fraction of halo mass) against radius in halo. Discounts extracentrals. Binned all together
def bh_mass_frac_vs_rad_noncentral(snapshot):
    with h5py.File("/orange/lblecha/pranavsatheesh/wandering_bhs/Data/wandering_bh_SM5_DFD_3_TNG.hdf5") as f:
        # is_wandering = f["bh/is_wandering"][:]
        mass = f["bh/mass_msun"] # all masses

        bh_start = f["snapshot/bh_start"][snapshot]
        num_bh = f["snapshot/n_bh"][snapshot]
        r_kpc = f["bh/r_kpc"][bh_start:bh_start+num_bh]
        mass = f["bh/r_kpc"][bh_start:bh_start+num_bh]
        
        for bh in range(num_bh):
            subhalo_row = f["bh/subhalo_row"][bh_start+bh]
            halo_mass = f["subhalo/subhalo_total_mass_msun"][subhalo_row]
            mass[bh] = mass[bh]/halo_mass
        valid_r = r_kpc[r_kpc>1e-8]
        valid_m = mass[r_kpc>1e-8]
        plt.scatter(np.log10(valid_r), np.log10(valid_m)) #, label=str(subhalochoice))
        # plt.axvline(np.log10(halfmass_rad), color="k") # better for individual halos
    #plt.legend()
    plt.xlabel("log10(radius [kpc])")
    plt.ylabel("log10(bh mass as fraction of halo)")
    plt.title("a") 
    plt.plot()
```
